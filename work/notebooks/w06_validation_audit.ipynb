### 1. Two Paper Findings & Methodology Questions

To practice rigorous academic reading, we examine two typical findings from search and ranking literature (such as FlyRank's research paper):

#### Finding 1: Semantic Embeddings Improve Relevance Ranking Over Keyword Overlap Alone
* **Methodology Question:** *Where does the ground-truth relevance label come from?*
* If labels are derived from historical user interactions (clicks or dwell time), are we controlling for presentation or position bias (where top-ranked items receive disproportionate clicks regardless of true relevance)? Furthermore, does the training-validation split ensure that query variations are completely isolated, or could overlapping queries between splits artificially boost performance metrics?

#### Finding 2: Multi-Feature Gradient Boosting Models Achieve Higher NDCG@10
* **Methodology Question:** *Does the validation design support the broader generalization claim?*
* If the validation split uses random shuffling across all available data rows, features built from global historical aggregates can leak information across fold boundaries. A more robust evaluation would require client-isolated or time-aware validation splits to prove the model generalizes to entirely unseen operational contexts.

### 3. Feature Leakage Audit

We inspect our final feature set to check for target leakage—specifically looking for features that incorporate post-prediction knowledge or future target statistics.

```python
# Feature Leakage Audit Simulation Code
def run_leakage_audit():
    # Example mock check of feature correlations against the target label
    features_inspected = [
        "historical_click_rate_30d",
        "query_term_match_count",
        "future_conversion_flag_leak",
        "item_popularity_score"
    ]

    correlations = [0.24, 0.41, 0.98, 0.19]

    leakage_df = pd.DataFrame({
        "Feature_Name": features_inspected,
        "Target_Correlation": correlations
    })

    # Flag anything suspiciously high (> 0.85 correlation)
    leakage_df["Flagged_As_Leakage"] = leakage_df["Target_Correlation"].abs() > 0.85

    print("--- LEAKAGE AUDIT REPORT ---")
    print(leakage_df.to_markdown(index=False))

run_leakage_audit()
```
